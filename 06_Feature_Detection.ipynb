## 6.1 What is a *feature*?

In notebooks 01–04, we worked with one spectrum at a time. In notebook 05, we saw that CascaDIA combined multiple spectra by combining MS2 scans with a fixed number of neighboring scans to form an augmented spectra. 

In feature detection, the goal is identify intact peptides across multiple MS1 spectra. Peptides elute on a slower time scale than mass spectrometers generates spectra, which means that peptides are detected in multiple MS1 scans. Feature detection algorithms attempt to locate all the MS1 peaks generated by a peptide across all MS1 scans. To answer that, we need to visualize the MS1 information. One way to do this is as a heatmap, a 2D image with m/z along the x-axis, retention time along the y-axis, and intensity represented by color. 

### The LC-MS map

During an LC-MS/MS run, the instrument collects MS1 scans as peptides elute off the liquid chromatography (LC) column. In our example data, MS1 scans were collected every second and peptides eluted over about 10s. While these numbers can vary depending on the mass spectrometer used and the LC condition, it is almost always the case that peptides elute more slowly than mass spectra are captured.  

If we stack MS1 scans on top of one another, ordered by retention time (i.e., the order that the scans were collected), we get a 2D map with three pieces of information:

- **Retention time (RT)**: when a peak was observed (which MS1 scan). These are the rows. This is discrete 
- **m/z**: where the peak sits on the mass axis. These are the columns. This is continuous
- **Intensity**: how much signal was observed at each (RT, m/z) position. This is the value in each cell.

```python
# The full LC-MS map of our example data: every MS1 peak in the file.
MZML_PATH = "Data/04-17-23_CA_Tryp_HCD_10min-calib.mzML"

# An unbounded RT and m/z window keeps every MS1 peak.
full_map = get_lcms_map_region(MZML_PATH, rt_range=(0, np.inf), mz_range=(0, np.inf))
print(f"{len(full_map):,} MS1 peaks across {full_map['scan'].nunique()} MS1 scans")
print(f"  RT range : {full_map['rt'].min():.2f} - {full_map['rt'].max():.2f} min")
print(f"  m/z range: {full_map['mz'].min():.1f} - {full_map['mz'].max():.1f}")

# Each row is one MS1 scan; each column is a 1 m/z bin; color is log10 intensity
# (MS1 intensities span several orders of magnitude, so a linear scale hides most peaks).
fig = plot_lcms_map(full_map, mz_bin=1.0)
fig.show()
```

### A peptide feature

A **peptide feature** is the complete MS1 signal from one peptide at one charge state. It has a characteristic shape along each axis:

- **Along m/z, an isotopic envelope.** A peptide doesn't produce a single peak. Elements like carbon and nitrogen have naturally occurring heavy isotopes (mostly ¹³C), so each peptide appears as a series of peaks: the monoisotopic peak (M), without any heavy isotopes incorporated, followed by the M+1, M+2, M+3... isotopologues. The monoisotopic peak is the tallest only for small peptides. Each extra residue adds another chance of carrying a heavy isotope, so above roughly 1.8 kDa the M+1 peak overtakes the monoisotopic peak, and for larger peptides the envelope's maximum keeps shifting to heavier isotopologues. Because we know that isotopologues are ~1 Dalton apart, the spacing between different isotopologue peaks enables us to calculate the charge state for each peptide. **`1/z`**  = the m/z spacing between subsequent peaks. A +2 ion has isotopes 0.5 m/z apart; a +3 ion, 0.333 m/z apart. 
- **Along RT, a bell-shaped elution profile.** A peptide elutes from the LC column over a few seconds to tens of seconds. Its intensity rises, reaches a maximum (the *apex*), and falls, giving an approximately Gaussian profile across consecutive MS1 scans.
- **Charge state.** Peptides are generally seen with charges of +1 to +8 charges. The charge is calculated based on the spacing between isotopologue peaks.

Put together, a feature is a small 2D blob on the map: a few isotope peaks wide, several scans tall, and brightest at its apex.

### Our example feature

Throughout this notebook, we'll use one feature from the same DDA file as notebooks 01–04. We chose it because it's clean and well resolved:

| Property | Value |

| Isolated precursor m/z | **1158.22** (the M+2 isotopologue, which the instrument selected for MS2) |
| Charge `z` | **+3** → isotope spacing `1/z ≈ 0.333` m/z |
| Retention time | **~13.06 min** (elution apex at MS1 scan 5450) |
| PSM scan | 5453 |

The envelope's peaks sit at 1157.55, 1157.88, 1158.22, 1158.55, 1158.88, and 1159.22 m/z. They are spaced ≈ 0.334 apart, so we know the charge state is +3. The first peak, 1157.55, is the monoisotopic peak, but it isn't the tallest: at ~3.5 kDa, this peptide is large enough that its M+1 and M+2 peaks are more intense (at the elution apex, the M+2 peak is ~1.7× the monoisotopic peak). DDA picks precursors by intensity, so the m/z the instrument recorded for the MS2 scan, 1158.22, is the M+2 isotopologue rather than the monoisotopic peak. Search engines such as MetaMorpheus correct for this by working back through the envelope to the monoisotopic mass before matching peptides. In the next cell, we load the region of the map around this feature (±0.15 min RT, ±1 m/z).

### Why detection is hard

Real runs are crowded. Many peptides co-elute, so features overlap: isotope envelopes interleave along m/z, and elution profiles overlap along RT.  For every point on the map, we need to decide which feature, if any, it belongs to. That makes feature detection a **segmentation** problem in 3D (RT, m/z, intensity) space.

Both models we'll look at frame detection this way. Where they differ is in how they encode the LC-MS map for the neural network:

- **DeepIso (2019)** renders the map as a fixed-resolution **grayscale image** and classifies pixels.
- **PointIso (2021)** keeps the map as a sparse **point cloud** of real peaks and segments the points directly.

As in notebooks 01–05, we treat the networks themselves (a CNN/RNN for DeepIso, PointNet with dual attention for PointIso) as black boxes and focus on the encoding. Section 6.2 builds the image, and Section 6.3 builds the point cloud and compares the two.

```python
# --- The chosen real feature (these constants are reused throughout the notebook) ---
# (MZML_PATH is defined in the LC-MS map cell above.)
FEATURE_MS2_SCAN = 5453        # PSM/MS2 scan number (from Data/ScanNumsWPsms.txt)
FEATURE_MZ       = 1158.2193   # isolated precursor m/z: the M+2 isotopologue the instrument picked for MS2
FEATURE_MONO_MZ  = 1157.5483   # monoisotopic peak m/z (two isotope spacings below FEATURE_MZ)
FEATURE_CHARGE   = 3           # charge z  ->  isotope spacing 1/z ~= 0.333 m/z
FEATURE_RT       = 13.065      # precursor retention time (min)

# LC-MS region around the feature: +/-0.15 min in RT, +/-1 m/z around the isolated peak.
# The m/z window spans the monoisotopic peak (1157.55) through the M+3 isotopologue (1159.22).
RT_WINDOW = (FEATURE_RT - 0.15, FEATURE_RT + 0.15)   # (12.915, 13.215) min
MZ_WINDOW = (FEATURE_MZ - 1.0, FEATURE_MZ + 1.0)     # (1157.219, 1159.219) m/z

# Load every real MS1 peak in that window: raw (scan, RT, m/z, intensity) points.
# Sections 6.2 and 6.3 encode these same points as an image and as a point cloud.
points = get_lcms_map_region(MZML_PATH, RT_WINDOW, MZ_WINDOW)


# Zoom in on our chosen feature using the same heatmap format as the full map above.
# With finer 0.05 m/z bins, the isotope envelope (peaks every 1/z ~= 0.333 m/z for z=3)
# separates into distinct vertical streaks, and the elution profile spans several scans.
# The box is the RT_WINDOW x MZ_WINDOW region we'll encode in Sections 6.2 and 6.3;
# the circle marks the isolated precursor (the M+2 peak at FEATURE_MZ). The monoisotopic
# peak (FEATURE_MONO_MZ) is the streak two isotope spacings (~0.67 m/z) to its left.
zoom = get_lcms_map_region(
    MZML_PATH,
    rt_range=(FEATURE_RT - 0.75, FEATURE_RT + 0.75),
    mz_range=(FEATURE_MZ - 8.0, FEATURE_MZ + 8.0),
)
fig = plot_lcms_map(
    zoom,
    mz_bin=0.05,
    title=f'LC-MS map around precursor {FEATURE_MZ} m/z (z={FEATURE_CHARGE}), RT {FEATURE_RT} min',
    highlight=dict(
        rt_range=RT_WINDOW,
        mz_range=MZ_WINDOW,
        point=(FEATURE_RT, FEATURE_MZ),
        label='feature window',
    ),
)
fig.show()
```

## 6.2 DeepIso: Encoding the map as an image

The map that we just plotted is essentially what DeepIso (Zohora et al., 2019) takes as input. Convolutional neural networks (CNNs) were designed to process images, and they expect a regular grid of pixels. DeepIso converts the LC-MS map into a grayscale image, then treats feature detection as an image-processing problem.

### Binning the map onto a fixed grid

DeepIso lays a fixed grid over the RT × m/z map and assigns every peak to a cell:

- **Rows are MS1 scans (`RT → 1 MS-scan`).** Each MS1 scan becomes one row, in acquisition order, with row 0 as the earliest scan in the window. RT is quantized to the instrument's scan rate.
- **Columns are 0.01 m/z bins (`m/z → 0.01`).** Each peak is assigned to the bin it falls in. If two peaks land in the same (scan, m/z bin) cell, their intensities are summed.
- **Cell values are intensities rescaled to 0–255.** Each cell becomes an 8-bit grayscale pixel, where 0 is empty and 255 is the most intense signal in the window.

After binning, the feature is an image. The isotope envelope appears as vertical streaks spaced `1/z ≈ 0.333` m/z (≈ 33 columns) apart, and the elution profile shows up as those streaks brightening and then fading from row to row.

### The [15 × 211] framing window

DeepIso doesn't classify the whole map at once. It slides a fixed **[15 scans × 211 m/z bins]** framing window across the image and labels the pixels inside each frame. The window is sized to fit one feature:

- **15 scans** spans one elution profile, since a feature is only a handful of scans wide at the apex.
- **211 bins** is `211 × 0.01 ≈ 2.11 m/z`, wide enough to hold a full isotope envelope for the charge states DeepIso targets.

We overlay this frame on the heatmap below to show how much of the map a single frame covers.

### The cost of fixed precision

Binning has a cost. Once we choose `m/z → 0.01` and `RT → 1 scan`, the resolution is fixed. Two isotopes closer than 0.01 m/z would fall into the same column, and the measured m/z (reported to four decimals) is rounded to its bin. The grid also covers every location, whether or not a peak was observed there, so most of the 15 × 211 ≈ 3,000 pixels in a frame are zero. PointIso addresses this in Section 6.3.

As in notebooks 01–05, we treat the DeepIso network (a CNN that scans each frame and an RNN that links detections across scans) as a black box. The next cell builds the image it takes as input.

```python
# DeepIso encoding: bin the LC-MS map region onto a fixed grid ->
#   rows  = MS1 scans (RT -> 1 scan)
#   cols  = 0.01-wide m/z bins (m/z -> 0.01)
#   value = summed intensity, squashed to an 8-bit grayscale pixel (0-255)
# The result is literally a small grayscale image of the feature.
image = build_lcms_image(points, mz_bin=0.01, mz_range=MZ_WINDOW)

grid = image['image']
n_scans, n_mz_bins = grid.shape
filled = int((grid > 0).sum())
print(f'DeepIso image: {n_scans} scans x {n_mz_bins} m/z bins = {grid.size} pixels')
print(f'  filled pixels: {filled} ({100 * filled / grid.size:.1f}%)  ->  the rest are empty grid')
print(f'  pixel range  : {int(grid.min())} - {int(grid.max())} (intensity {image["max_intensity"]:.3g} -> 255)')

# Show the image as an RT x m/z heatmap, overlaying the DeepIso [15 scans x 211 m/z bins]
# framing window (centered on the brightest pixel) so you can see one "look" of the model.
fig = plot_lcms_heatmap(
    image,
    title='DeepIso image of the feature (0.01 m/z bins) - [15 x 211] frame overlaid',
    frame=True,
)
fig.show()
```

## 6.3 PointIso: Encoding the map as a point cloud

The DeepIso image works, but it has a cost. To feed a CNN, we laid a dense grid over the map, quantized m/z to 0.01 and RT to one scan, rescaled every intensity to an 8-bit pixel, and stored a cell for every (scan, m/z bin) location, even though almost all of them are empty. PointIso (Zohora et al., 2021) drops the grid entirely. Why store empty space at all?

### The (RT, m/z, intensity) point cloud

An LC-MS map is really a sparse list of peaks. PointIso encodes it that way, as a **point cloud** where each detected MS1 peak is a single point with three coordinates:

$$\text{point} = (\text{RT},\; m/z,\; \text{intensity}).$$

There is no grid and no empty cells. The map is just the set of peaks that were measured. Computer graphics uses the same representation for 3D shapes (a cloud of surface points), which is why PointIso can use **PointNet**, a network designed to take point clouds as input.

### Arbitrary precision

Each point stores its own coordinates, so there's no bin to round to. The measured m/z keeps its full precision (up to four decimals), and RT and intensity are kept as recorded. The `1/z ≈ 0.333` m/z isotope spacing, which DeepIso approximated as ≈ 33 columns, is preserved at whatever resolution the instrument delivered. Nothing is quantized, summed into a shared cell, or rescaled to 0–255, so the encoding doesn't impose a resolution limit on the data.

### A non-overlapping window

DeepIso's [15 × 211] frame overlaps itself as it slides across the image, so each pixel is examined in several frames. PointIso instead divides the map into **2.0 m/z × 15 scan** windows that tile it without overlap, so each region is examined once. Like DeepIso's frame, the window is sized to hold one isotope envelope (2.0 m/z covers the isotopes for the charge states PointIso targets) and one elution profile (15 scans). Because each point carries its own coordinates, the window doesn't need an internal grid.

### Still a segmentation problem

The task hasn't changed from Section 6.1: for every point, decide which feature it belongs to. On a point cloud, this is **point-wise segmentation**. PointNet, combined with a dual-attention module, assigns each point to a feature or to background, and a second stage groups the segmented points into charge-resolved features. We'll treat this network as a black box as well.

The next cell builds the point cloud for the same feature we imaged in Section 6.2. After that, we compare the two encodings side by side, counting filled vs. empty pixels and bytes, and close with the 4D ion-mobility extension, which the point cloud handles and the image doesn't.

```python
# PointIso encoding: keep the LC-MS map region as a sparse point cloud ->
#   each real MS1 peak = one (RT, m/z, intensity) triplet at its measured coordinates.
# No grid, no binning, no 0-255 rescaling -- full precision, only real peaks stored.
cloud = lcms_point_cloud(points, mz_range=MZ_WINDOW)

print(f'PointIso point cloud: {len(cloud)} real peaks, each a (RT, m/z, intensity) triplet')
print(f'  m/z kept to full precision (e.g. {cloud[0, 1]:.4f}) -- no 0.01 bin snapping')
print(f'  RT range : {cloud[:, 0].min():.3f} - {cloud[:, 0].max():.3f} min')
print(f'  m/z range: {cloud[:, 1].min():.4f} - {cloud[:, 1].max():.4f}')

# Show the same feature as an interactive 3D scatter of its real peaks. Rotate to see the
# isotope envelope along m/z (spacing 1/z ~= 0.333) and the elution bell along RT -- the
# same structure as the DeepIso image, but stored as points instead of a dense grid.
fig = plot_lcms_point_cloud(
    cloud,
    title=f'PointIso point cloud of the feature - precursor {FEATURE_MZ} m/z (z={FEATURE_CHARGE})',
)
fig.show()
```

### Image vs. point cloud: Sparsity and memory

We now have the same feature encoded two ways: the DeepIso image (`image`, from Section 6.2) and the PointIso point cloud (`cloud`, from Section 6.3).

An LC-MS window is almost entirely empty. A peptide occupies a handful of isotope peaks over a few scans, and the rest of the window is background. The point cloud stores only the real peaks, so its size depends on the data. The image stores a cell for every (scan, m/z bin) location, so its size depends on the chosen resolution, regardless of how much signal is present.

The next cell measures two consequences of this on our feature:

- **Sparsity.** Even at DeepIso's 0.01 m/z bins, the vast majority of the image's pixels are zero. The point cloud has no empty entries.
- **Resolution cost.** The image's size is tied to its bin width. Each additional decimal place of m/z resolution multiplies the pixel count by 10: going from `0.01 → 0.001 → 0.0001` takes a ~2 m/z envelope from thousands of pixels toward the paper's ~300,000. The point cloud stays the same size, because it stores the same real peaks at full precision regardless of resolution.

The image has to pick a bin, so it loses precision. The point cloud stores coordinates, so it keeps it. This is also why the point cloud scales to the 4D extension below.

```python
# Side-by-side: the SAME window (`image` from Section 6.2, `cloud` from Section 6.3),
# counted two ways -- filled vs. empty pixels, and bytes on disk.
fp = lcms_encoding_footprint(image, cloud)

print("Same LC-MS window, two encodings")
print("=" * 46)
print(f"{'':22}{'DeepIso image':>14}{'':>2}{'PointIso cloud':>0}")
print(f"{'stored cells / points':22}{fp['n_pixels']:>14,}{'':>2}{fp['n_points']:>14,}")
print(f"{'  of which real peaks':22}{fp['n_filled']:>14,}{'':>2}{fp['n_points']:>14,}")
print(f"{'  empty grid':22}{fp['n_empty']:>14,}{'':>2}{0:>14,}")
print(f"{'sparsity (fraction 0)':22}{fp['sparsity']:>13.1%}{'':>3}{'--':>14}")
print(f"{'bytes':22}{fp['image_bytes']:>14,}{'':>2}{fp['point_cloud_bytes']:>14,}")
print("=" * 46)
print(f"At 0.01 m/z bins the image is {fp['sparsity']:.1%} empty and costs "
      f"{fp['memory_ratio']:.1f}x the point cloud's memory.")
print()

# Now shrink the m/z bin toward the paper's fine resolution. The image grows
# 10x per decimal (a cell exists everywhere), while the point cloud never
# changes -- it only ever stores the real peaks.
print("As we ask the image for finer m/z resolution:")
print(f"{'m/z bin':>10}{'image pixels':>16}{'image bytes':>14}{'sparsity':>11}{'vs cloud':>11}")
for mz_bin in (0.01, 0.001, 0.0001):
    img_b = build_lcms_image(points, mz_bin=mz_bin, mz_range=MZ_WINDOW)
    fpb = lcms_encoding_footprint(img_b, cloud)
    print(f"{mz_bin:>10}{fpb['n_pixels']:>16,}{fpb['image_bytes']:>14,}"
          f"{fpb['sparsity']:>10.2%}{fpb['memory_ratio']:>10.0f}x")
print(f"{'point cloud':>10}{fp['n_points']:>16,}{fp['point_cloud_bytes']:>14,}"
      f"{'--':>11}{'1x':>11}  (unchanged at every resolution)")
```

### The 4D extension: Adding ion mobility

The gap between the two encodings widens when the instrument adds a dimension. On a **timsTOF** (trapped ion mobility spectrometry, TIMS), ions are separated by m/z, retention time, and **ion mobility**: how quickly an ion drifts through a gas, which depends on its collision cross section (shape and size). Mobility is reported as the inverse reduced ion mobility, `1/K₀`. A peptide feature now spreads across m/z, RT, and `1/K₀`, which separates co-eluting, overlapping ions better than m/z and RT alone.

The two encodings handle this extra dimension very differently:

- **Point cloud (PointIso).** Each peak goes from a 3-tuple `(RT, m/z, intensity)` to a 4-tuple `(RT, m/z, intensity, 1/K₀)`. Each point gets one more number, and the number of points is still the number of real peaks, so the encoding's size still tracks the data. This is the 4D extension PointIso targets.
- **Image (DeepIso).** The 2D grid becomes a 3D volume, with every `(scan, m/z bin)` pixel repeated across a new axis of `1/K₀` bins. The pixel count is multiplied by the number of mobility bins, so memory grows multiplicatively with each added dimension. The map was already ~99% empty in 2D, and the added dimension makes the volume emptier still. A grid pays for resolution in every dimension, whether or not there's signal there.

The image's cost scales with resolution and dimensionality, while the point cloud's cost scales with the data. That's why the sparse encoding generalizes from 3D LC-MS maps to 4D ion-mobility data, and to any dimension an instrument adds next.

## Summary

In this notebook, we moved from the single MS2 spectra of notebooks 01–05 to the **LC-MS map** (MS1 signal over retention time) and looked at how two feature detection models encode that map for a neural network.

1. **A peptide feature is a shape on the map.** One peptide at one charge state produces an isotope envelope spaced `1/z` in m/z and a bell-shaped elution profile over RT. Real runs are crowded with overlapping features, so detection is framed as **3D segmentation**: deciding which feature each point belongs to.

2. **DeepIso encodes the map as a fixed-resolution image.** It bins RT to 1 scan and m/z to 0.01, rescales intensity to a 0–255 pixel, and slides a [15 × 211] frame across the grid. A CNN can read the result like a photograph, but the resolution is fixed (exact m/z is rounded to its bin), and a cell exists at every location, so the image is ~99% empty.

3. **PointIso encodes the map as a point cloud.** It keeps every real peak as an `(RT, m/z, intensity)` triplet at full precision, with no grid, binning, or rounding. The same feature takes a fraction of the memory, and PointNet segments the points directly.

The image's cost scales with **resolution and dimensionality**, while the point cloud's cost scales with the **data**. That's why the point cloud keeps full precision, stores no empty space, and extends to 4D timsTOF ion-mobility data, where the image would become a mostly empty volume. As in the other notebooks, we treated both networks as black boxes and focused on the encoding.

## Appendix: Function Definitions

Below are the definitions of key utility functions used throughout this notebook. They live in `util.py` and are imported at the top, so you don't need to run these cells for the notebook to work — they're here purely for reference. Feel free to skip this section if you're not interested in the implementation details.

- **`get_lcms_map_region`** reads the raw `(scan, RT, m/z, intensity)` points of an LC-MS map region from the mzML file. Both encodings start from its output. It relies on `extract_scan_number` and `pyteomics`, also in `util.py`.
- **`plot_lcms_map`** draws a whole LC-MS map (all MS1 scans) as an RT × m/z heatmap with coarse m/z bins and a log10 intensity color scale.
- **`build_lcms_image`** and **`plot_lcms_heatmap`** implement the DeepIso image encoding: they bin the map onto a fixed RT × m/z grid of 0–255 pixels and draw it as a heatmap, optionally with the [15 × 211] framing window.
- **`lcms_point_cloud`** and **`plot_lcms_point_cloud`** implement the PointIso point-cloud encoding: they keep every real peak as an `(RT, m/z, intensity)` triplet at full precision and draw it as a 3D scatter (or as sticks rising from the m/z × RT plane).
- **`lcms_encoding_footprint`** compares the two encodings on sparsity and memory (filled vs. empty pixels, image bytes vs. point-cloud bytes).

```python
def get_lcms_map_region(mzml_path, rt_range, mz_range):
    """Extract the LC-MS map region for a peptide feature as a flat point list.

    Walks the MS1 scans of an mzML file and collects every peak whose retention
    time falls in ``rt_range`` and whose m/z falls in ``mz_range``. This is the
    raw (retention time, m/z, intensity) representation of an LC-MS map region --
    the shared starting point for both the DeepIso image encoding and the
    PointIso point-cloud encoding in notebook 06.

    Parameters
    ----------
    mzml_path : str
        Path to the mzML file (e.g. the committed DDA calibration file).
    rt_range : (float, float)
        Inclusive (low, high) retention-time window in minutes.
    mz_range : (float, float)
        Inclusive (low, high) m/z window.

    Returns
    -------
    pandas.DataFrame
        One row per peak with columns ``scan`` (int MS1 scan number),
        ``rt`` (float, minutes), ``mz`` (float), and ``intensity`` (float),
        sorted by ``rt`` then ``mz``. Empty (zero-row) frame if nothing matches.
    """
    lo_rt, hi_rt = rt_range
    lo_mz, hi_mz = mz_range

    scans, rts, mzs, intensities = [], [], [], []
    with pyteomics.mzml.read(mzml_path) as spectra:
        for spectrum in spectra:
            if spectrum.get('ms level', 0) != 1:
                continue
            rt_raw = spectrum['scanList']['scan'][0]['scan start time']
            rt = float(rt_raw)
            # mzML may store RT in seconds (e.g. files written by pyopenms); work in minutes.
            if getattr(rt_raw, 'unit_info', None) == 'second':
                rt /= 60.0
            if rt < lo_rt or rt > hi_rt:
                continue
            scan = extract_scan_number(spectrum['id'])
            mz = spectrum['m/z array']
            intensity = spectrum['intensity array']
            mask = (mz >= lo_mz) & (mz <= hi_mz)
            n = int(mask.sum())
            if n == 0:
                continue
            scans.append(np.full(n, scan, dtype=int))
            rts.append(np.full(n, rt))
            mzs.append(mz[mask])
            intensities.append(intensity[mask])

    if not scans:
        return pd.DataFrame(columns=['scan', 'rt', 'mz', 'intensity'])

    df = pd.DataFrame({
        'scan': np.concatenate(scans),
        'rt': np.concatenate(rts),
        'mz': np.concatenate(mzs),
        'intensity': np.concatenate(intensities),
    })
    return df.sort_values(['rt', 'mz']).reset_index(drop=True)
```

```python
def plot_lcms_map(points, mz_bin=1.0, title=None, highlight=None):
    """Plot a whole LC-MS map as an RT x m/z heatmap of log10 intensity.

    This is the "big picture" view of the MS1 data: every MS1 scan is one row
    (retention time up the y-axis), m/z runs along the x-axis in ``mz_bin``-wide
    bins, and color is the log10 of the summed intensity in each cell. The
    orientation matches DeepIso (Zohora et al., 2019) and :func:`plot_lcms_heatmap`:
    rows are MS1 scans, columns are m/z bins. Unlike
    :func:`plot_lcms_heatmap` (a DeepIso-style 0.01 m/z, 0-255 image of one small
    window), this view is meant for looking at an entire run, so it uses coarse m/z
    bins and a log color scale -- MS1 intensities span several orders of magnitude,
    and on a linear scale only the few brightest features would be visible.

    Parameters
    ----------
    points : pandas.DataFrame
        Output of :func:`get_lcms_map_region` -- rows of ``scan``, ``rt``, ``mz``,
        ``intensity``. Pass a very wide window to get the full map.
    mz_bin : float, optional
        Width of each m/z bin (column) in m/z units. Default ``1.0``.
    title : str, optional
        Plot title. Defaults to a description of the map size.
    highlight : dict, optional
        Mark a feature on the map. Recognized keys (all optional):

        * ``rt_range`` and ``mz_range`` -- (low, high) tuples; draws a box.
        * ``point`` -- (rt, mz) tuple; draws an open-circle marker (e.g. the precursor).
        * ``label`` -- text placed at the top-left corner of the box (or next to the point).

    Returns
    -------
    plotly.graph_objs.Figure or matplotlib.figure.Figure
        Interactive Plotly heatmap if Plotly is available, else a matplotlib figure.
    """
    highlight = highlight or {}
    box = None  # in plot coordinates: x = m/z, y = RT
    if 'rt_range' in highlight and 'mz_range' in highlight:
        (y0, y1), (x0, x1) = highlight['rt_range'], highlight['mz_range']
        box = dict(x0=x0, x1=x1, y0=y0, y1=y1)
    point = highlight.get('point')  # given as (rt, mz); drawn at x = mz, y = rt
    label = highlight.get('label')
    hl_color = '#FF4081'  # magenta: stands out against Viridis and the black background

    # Rows: one per MS1 scan, in RT order.
    rt_values, row_idx = np.unique(points['rt'].to_numpy(dtype=float), return_inverse=True)
    # Columns: fixed-width m/z bins.
    mz = points['mz'].to_numpy(dtype=float)
    lo_mz = np.floor(mz.min() / mz_bin) * mz_bin
    n_mz = int(np.ceil((mz.max() - lo_mz) / mz_bin)) + 1
    col_idx = np.clip(((mz - lo_mz) / mz_bin).astype(int), 0, n_mz - 1)
    mz_centers = lo_mz + (np.arange(n_mz) + 0.5) * mz_bin

    grid = np.zeros((len(rt_values), n_mz))
    np.add.at(grid, (row_idx, col_idx), points['intensity'].to_numpy(dtype=float))
    log_grid = np.where(grid > 0, np.log10(np.where(grid > 0, grid, 1.0)), np.nan)

    if title is None:
        title = (f'LC-MS map: {len(rt_values)} MS1 scans, '
                 f'{mz_centers[0] - mz_bin / 2:.0f}-{mz_centers[-1] + mz_bin / 2:.0f} m/z '
                 f'({mz_bin:g} m/z bins)')

    try:
        import plotly.graph_objects as go
        fig = go.Figure(data=go.Heatmap(
            z=log_grid,
            x=mz_centers,
            y=rt_values,
            colorscale='Viridis',
            colorbar=dict(title='log10<br>intensity'),
            hovertemplate=('m/z: %{x:' + ('.1f' if mz_bin >= 0.1 else '.2f')
                           + '}<br>RT: %{y:.3f} min<br>log10 intensity: %{z:.2f}<extra></extra>'),
        ))
        if box is not None:
            fig.add_shape(type='rect', **box, line=dict(color=hl_color, width=2),
                          fillcolor='rgba(0,0,0,0)')
        if point is not None:
            fig.add_trace(go.Scatter(
                x=[point[1]], y=[point[0]], mode='markers', showlegend=False,
                marker=dict(symbol='circle-open', size=14, color=hl_color, line=dict(width=2)),
                hovertemplate='m/z: %{x:.4f}<br>RT: %{y:.3f} min<extra>highlight</extra>',
            ))
        if label:
            lx, ly = (box['x0'], box['y1']) if box is not None else (point[1], point[0])
            fig.add_annotation(x=lx, y=ly, text=label, showarrow=False,
                               xanchor='left', yanchor='bottom',
                               font=dict(color=hl_color, size=12))
        fig.update_layout(
            title=title,
            xaxis_title='m/z',
            yaxis_title='Retention time (min)',
            plot_bgcolor='black',
            height=600,
        )
        return fig
    except Exception:
        import logging
        logging.getLogger(__name__).warning(
            'Plotly unavailable for plot_lcms_map; falling back to matplotlib'
        )
        import matplotlib.pyplot as plt
        fig, ax = plt.subplots(figsize=(9, 6))
        ax.set_facecolor('black')
        mesh = ax.pcolormesh(mz_centers, rt_values, log_grid, cmap='viridis', shading='nearest')
        fig.colorbar(mesh, ax=ax, label='log10 intensity')
        if box is not None:
            import matplotlib.patches as mpatches
            ax.add_patch(mpatches.Rectangle(
                (box['x0'], box['y0']), box['x1'] - box['x0'], box['y1'] - box['y0'],
                fill=False, edgecolor=hl_color, linewidth=2,
            ))
        if point is not None:
            ax.plot(point[1], point[0], 'o', mfc='none', mec=hl_color, ms=12, mew=2)
        if label:
            lx, ly = (box['x0'], box['y1']) if box is not None else (point[1], point[0])
            ax.text(lx, ly, label, color=hl_color, va='bottom', ha='left')
        ax.set_xlabel('m/z')
        ax.set_ylabel('Retention time (min)')
        ax.set_title(title)
        fig.tight_layout()
        return fig
```

```python
def build_lcms_image(points, mz_bin=0.01, mz_range=None, max_intensity=None):
    """Encode an LC-MS map region as a fixed-resolution 2D grayscale image (DeepIso).

    This is the DeepIso (Zohora et al., 2019) encoding: the LC-MS map is binned
    onto a fixed grid and every cell's intensity is squashed to a 0-255 grayscale
    "pixel", exactly as a black-and-white image. The two axes use fixed precision:

    * **RT axis (rows):** one row per MS1 scan (``RT -> 1 MS-scan``). Rows are
      ordered by increasing retention time, so row 0 is the earliest scan.
    * **m/z axis (columns):** fixed ``mz_bin`` bins (0.01 m/z in DeepIso). Peaks
      landing in the same (scan, m/z-bin) cell have their intensities summed.

    Fixing the resolution is the whole point of the comparison with PointIso: it
    forces a grid whether or not real peaks are present, so most pixels end up
    empty (see :func:`lcms_encoding_footprint`).

    Parameters
    ----------
    points : pandas.DataFrame
        Output of :func:`get_lcms_map_region` -- rows of ``scan``, ``rt``, ``mz``,
        ``intensity``.
    mz_bin : float, optional
        Width of each m/z bin (column) in m/z units. Default ``0.01`` (DeepIso).
    mz_range : (float, float), optional
        Inclusive (low, high) m/z span for the columns. Defaults to the min/max
        m/z of ``points``. Peaks outside the range are dropped.
    max_intensity : float, optional
        Intensity mapped to pixel value 255. Defaults to the max intensity in
        ``points``. Pass an explicit value to scale several windows consistently.

    Returns
    -------
    dict
        ``image`` : 2D ``uint8`` array, shape ``(n_scans, n_mz_bins)``, 0-255.
        ``rt_values`` : 1D array, the RT (min) of each row.
        ``scan_values`` : 1D int array, the MS1 scan number of each row.
        ``mz_edges`` : 1D array of length ``n_mz_bins + 1``, the m/z bin edges.
        ``mz_bin`` : the bin width used.
        ``max_intensity`` : the intensity mapped to 255.
    """
    cols = ['scan', 'rt', 'mz', 'intensity']
    if points is None or len(points) == 0:
        return {
            'image': np.zeros((0, 0), dtype=np.uint8),
            'rt_values': np.zeros(0),
            'scan_values': np.zeros(0, dtype=int),
            'mz_edges': np.zeros(0),
            'mz_bin': mz_bin,
            'max_intensity': 0.0,
        }
    df = points[cols]

    if mz_range is None:
        lo_mz, hi_mz = float(df['mz'].min()), float(df['mz'].max())
    else:
        lo_mz, hi_mz = float(mz_range[0]), float(mz_range[1])
        df = df[(df['mz'] >= lo_mz) & (df['mz'] <= hi_mz)]

    # m/z columns: fixed-width bins spanning [lo_mz, hi_mz].
    n_mz = max(1, int(np.ceil((hi_mz - lo_mz) / mz_bin)))
    mz_edges = lo_mz + mz_bin * np.arange(n_mz + 1)

    # RT rows: one per MS1 scan, ordered by retention time.
    scan_order = df[['scan', 'rt']].drop_duplicates().sort_values('rt')
    scan_values = scan_order['scan'].to_numpy(dtype=int)
    rt_values = scan_order['rt'].to_numpy(dtype=float)
    row_of_scan = {s: i for i, s in enumerate(scan_values)}
    n_scans = len(scan_values)

    grid = np.zeros((n_scans, n_mz), dtype=float)
    if len(df):
        rows = df['scan'].map(row_of_scan).to_numpy(dtype=int)
        col = np.floor((df['mz'].to_numpy() - lo_mz) / mz_bin).astype(int)
        col = np.clip(col, 0, n_mz - 1)
        np.add.at(grid, (rows, col), df['intensity'].to_numpy())

    if max_intensity is None:
        max_intensity = float(grid.max()) if grid.size else 0.0
    if max_intensity > 0:
        image = np.clip(np.round(grid / max_intensity * 255.0), 0, 255).astype(np.uint8)
    else:
        image = np.zeros_like(grid, dtype=np.uint8)

    return {
        'image': image,
        'rt_values': rt_values,
        'scan_values': scan_values,
        'mz_edges': mz_edges,
        'mz_bin': mz_bin,
        'max_intensity': max_intensity,
    }
```

```python
def lcms_point_cloud(points, mz_range=None):
    """Encode an LC-MS map region as a (RT, m/z, intensity) point cloud (PointIso).

    This is the PointIso (Zohora et al., 2021) encoding: instead of binning the
    LC-MS map onto a fixed grid (see :func:`build_lcms_image`), every real peak is
    kept as a single ``(retention time, m/z, intensity)`` triplet at its measured
    coordinates. There is **no binning** -- the m/z and RT axes keep their full
    precision, and only positions where a peak was actually observed are stored.

    That is the whole contrast with the DeepIso image: the image spends memory on
    a dense grid of mostly-empty pixels at a fixed resolution, while the point
    cloud stores only the handful of real peaks at arbitrary precision (and, in the
    4D TimsTOF extension, extends naturally to a fourth ion-mobility coordinate).
    See :func:`lcms_encoding_footprint` for the side-by-side sparsity/memory story.

    Parameters
    ----------
    points : pandas.DataFrame
        Output of :func:`get_lcms_map_region` -- rows of ``scan``, ``rt``, ``mz``,
        ``intensity``.
    mz_range : (float, float), optional
        Inclusive (low, high) m/z span. Defaults to the full range of ``points``.
        Peaks outside the range are dropped, matching :func:`build_lcms_image` so
        the two encodings describe exactly the same window.

    Returns
    -------
    numpy.ndarray
        Float array of shape ``(n_points, 3)`` with columns ``[rt, mz,
        intensity]``, one row per real peak, sorted by ``rt`` then ``mz``. Empty
        ``(0, 3)`` array if ``points`` is empty.
    """
    if points is None or len(points) == 0:
        return np.zeros((0, 3), dtype=float)

    df = points[['rt', 'mz', 'intensity']]
    if mz_range is not None:
        lo_mz, hi_mz = float(mz_range[0]), float(mz_range[1])
        df = df[(df['mz'] >= lo_mz) & (df['mz'] <= hi_mz)]

    if len(df) == 0:
        return np.zeros((0, 3), dtype=float)

    df = df.sort_values(['rt', 'mz'])
    return df.to_numpy(dtype=float)
```

```python
def lcms_encoding_footprint(image, cloud):
    """Compare the DeepIso image and PointIso point cloud on sparsity and memory.

    This is the core takeaway of notebook 06, quantified: for the *same* LC-MS
    window, how much of the fixed-resolution image is actually empty, and how many
    bytes does each encoding cost? The image (see :func:`build_lcms_image`) spends
    one pixel per (scan, m/z-bin) cell whether or not a peak is there; the point
    cloud (see :func:`lcms_point_cloud`) stores only real peaks. The punchline is
    that a high-resolution window is ~300,000 pixels but only a few thousand real
    points, so the vast majority of the image is empty space.

    Parameters
    ----------
    image : dict
        Output of :func:`build_lcms_image` (uses the ``image`` 2D ``uint8`` array).
    cloud : numpy.ndarray
        Output of :func:`lcms_point_cloud` -- ``(n_points, 3)`` float array.

    Returns
    -------
    dict
        ``n_pixels`` : total grid cells in the image (n_scans x n_mz_bins).
        ``n_filled`` : image cells with a non-zero pixel (a real peak landed there).
        ``n_empty`` : image cells that are zero (wasted on empty space).
        ``fill_fraction`` : ``n_filled / n_pixels`` (0-1); ``sparsity`` is 1 minus it.
        ``sparsity`` : fraction of the image that is empty.
        ``n_points`` : number of real peaks in the point cloud.
        ``image_bytes`` : bytes to store the dense ``uint8`` image grid.
        ``point_cloud_bytes`` : bytes to store the ``(n_points, 3)`` float triplets.
        ``memory_ratio`` : ``image_bytes / point_cloud_bytes`` (image / point cloud).
    """
    grid = np.asarray(image['image']) if isinstance(image, dict) else np.asarray(image)
    cloud = np.asarray(cloud, dtype=float)
    if cloud.ndim != 2 or cloud.shape[1] != 3:
        cloud = cloud.reshape(-1, 3) if cloud.size else np.zeros((0, 3))

    n_pixels = int(grid.size)
    n_filled = int(np.count_nonzero(grid))
    n_empty = n_pixels - n_filled
    fill_fraction = (n_filled / n_pixels) if n_pixels else 0.0

    n_points = int(len(cloud))
    image_bytes = int(grid.nbytes)
    point_cloud_bytes = int(cloud.nbytes)
    memory_ratio = (image_bytes / point_cloud_bytes) if point_cloud_bytes else float('inf')

    return {
        'n_pixels': n_pixels,
        'n_filled': n_filled,
        'n_empty': n_empty,
        'fill_fraction': fill_fraction,
        'sparsity': 1.0 - fill_fraction,
        'n_points': n_points,
        'image_bytes': image_bytes,
        'point_cloud_bytes': point_cloud_bytes,
        'memory_ratio': memory_ratio,
    }
```

```python
def plot_lcms_heatmap(image, title=None, frame=None):
    """Plot a DeepIso LC-MS image as an RT x m/z heatmap (intensity -> pixel 0-255).

    Renders the fixed-resolution grid produced by :func:`build_lcms_image`: each
    row is one MS1 scan (RT axis), each column is a fixed m/z bin, and the cell
    value is the 0-255 grayscale "pixel". Retention time (minutes) is used for the
    y-axis and m/z-bin centers for the x-axis so the picture is readable, while the
    pixel values are exactly what the DeepIso CNN would ingest.

    Optionally overlays the DeepIso **[15 scans x 211 m/z bins]** framing window as a
    rectangle, so the notebook can show how the model slides a fixed frame across
    the map. Prefers Plotly (interactive), falling back to matplotlib -- matching
    the rest of ``util.py``.

    Parameters
    ----------
    image : dict
        Output of :func:`build_lcms_image` (keys ``image``, ``rt_values``,
        ``mz_edges``, ``mz_bin``, ...).
    title : str, optional
        Plot title. Defaults to a description of the window size.
    frame : bool or dict, optional
        Overlay the DeepIso framing window. ``True`` draws a default
        [15 x 211] frame centered on the brightest pixel. A dict may override:

        * ``n_scans`` (int, default 15) -- frame height in MS1 scans (rows).
        * ``n_mz_bins`` (int, default 211) -- frame width in m/z bins (columns).
        * ``row0`` (int) -- top row of the frame; defaults to centering on the
          brightest pixel's row.
        * ``col0`` (int) -- left column of the frame; defaults to centering on the
          brightest pixel's column.

    Returns
    -------
    plotly.graph_objs.Figure or matplotlib.figure.Figure
        Interactive Plotly heatmap if Plotly is available, else a matplotlib figure.
    """
    grid = np.asarray(image['image'])
    rt_values = np.asarray(image['rt_values'], dtype=float)
    mz_edges = np.asarray(image['mz_edges'], dtype=float)
    mz_bin = float(image.get('mz_bin', 0.01))
    n_scans, n_mz = grid.shape if grid.ndim == 2 else (0, 0)
    mz_centers = (mz_edges[:-1] + mz_edges[1:]) / 2 if len(mz_edges) >= 2 else np.zeros(0)

    if title is None:
        title = f'DeepIso LC-MS image ({n_scans} scans x {n_mz} m/z bins @ {mz_bin} m/z)'

    # Resolve the optional [n_scans x n_mz_bins] framing window into a rectangle.
    frame_rect = None
    if frame and n_scans and n_mz:
        opts = frame if isinstance(frame, dict) else {}
        fr_scans = int(opts.get('n_scans', 15))
        fr_bins = int(opts.get('n_mz_bins', 211))
        if grid.size:
            peak_row, peak_col = np.unravel_index(int(np.argmax(grid)), grid.shape)
        else:
            peak_row, peak_col = 0, 0
        row0 = int(opts.get('row0', peak_row - fr_scans // 2))
        col0 = int(opts.get('col0', peak_col - fr_bins // 2))
        row0 = int(np.clip(row0, 0, max(0, n_scans - 1)))
        col0 = int(np.clip(col0, 0, max(0, n_mz - 1)))
        row1 = min(n_scans - 1, row0 + fr_scans - 1)
        col1 = min(n_mz - 1, col0 + fr_bins - 1)
        frame_rect = {
            'y0': rt_values[row0], 'y1': rt_values[row1],
            'x0': mz_edges[col0], 'x1': mz_edges[col1 + 1],
            'n_scans': fr_scans, 'n_mz_bins': fr_bins,
        }

    try:
        import plotly.graph_objects as go
        fig = go.Figure(data=go.Heatmap(
            z=grid,
            x=mz_centers,
            y=rt_values,
            colorscale='Greys',
            reversescale=True,
            zmin=0, zmax=255,
            colorbar=dict(title='pixel (0-255)'),
            hovertemplate='m/z: %{x:.3f}<br>RT: %{y:.3f} min<br>pixel: %{z}<extra></extra>',
        ))
        if frame_rect is not None:
            fig.add_shape(
                type='rect',
                x0=frame_rect['x0'], x1=frame_rect['x1'],
                y0=frame_rect['y0'], y1=frame_rect['y1'],
                line=dict(color='#D32F2F', width=2),
                fillcolor='rgba(0,0,0,0)',
            )
            fig.add_annotation(
                x=frame_rect['x1'], y=frame_rect['y1'],
                text=f"[{frame_rect['n_scans']} x {frame_rect['n_mz_bins']}] frame",
                showarrow=False, font=dict(color='#D32F2F', size=11),
                xanchor='right', yanchor='bottom',
            )
        fig.update_layout(
            title=title,
            xaxis_title='m/z',
            yaxis_title='Retention time (min)',
            plot_bgcolor='white',
        )
        return fig
    except Exception:
        import logging
        logging.getLogger(__name__).warning(
            'Plotly unavailable for plot_lcms_heatmap; falling back to matplotlib'
        )
        import matplotlib.pyplot as plt
        fig, ax = plt.subplots(figsize=(8, 6))
        if grid.size and len(mz_edges) >= 2 and len(rt_values):
            extent = [mz_edges[0], mz_edges[-1], rt_values[-1], rt_values[0]]
            ax.imshow(grid, aspect='auto', cmap='gray_r', vmin=0, vmax=255,
                      extent=extent, interpolation='nearest')
        if frame_rect is not None:
            import matplotlib.patches as mpatches
            ax.add_patch(mpatches.Rectangle(
                (frame_rect['x0'], min(frame_rect['y0'], frame_rect['y1'])),
                frame_rect['x1'] - frame_rect['x0'],
                abs(frame_rect['y1'] - frame_rect['y0']),
                fill=False, edgecolor='#D32F2F', linewidth=2,
            ))
        ax.set_xlabel('m/z')
        ax.set_ylabel('Retention time (min)')
        ax.set_title(title)
        fig.tight_layout()
        return fig
```

```python
def plot_lcms_point_cloud(cloud, title=None, sticks=False):
    """Plot a PointIso LC-MS point cloud as a 3D (m/z, RT, intensity) scatter.

    Renders the arbitrary-precision triplets produced by
    :func:`lcms_point_cloud`: one marker per real peak at its measured
    ``(m/z, retention time, intensity)``, with no binning. Markers are colored by
    intensity. This is the visual counterpart to :func:`plot_lcms_heatmap` -- the
    same feature, stored as sparse points instead of a dense grid. Prefers Plotly
    (interactive 3D), falling back to a matplotlib 3D scatter.

    Parameters
    ----------
    cloud : numpy.ndarray
        ``(n_points, 3)`` array of ``[rt, mz, intensity]`` rows, as returned by
        :func:`lcms_point_cloud`.
    title : str, optional
        Plot title. Defaults to a description including the point count.
    sticks : bool, optional
        If True, draw each peak as a vertical stick rising from the m/z x RT plane
        (intensity 0) up to its intensity -- the familiar centroided-spectrum look,
        repeated across every MS1 scan. Default False (markers only).

    Returns
    -------
    plotly.graph_objs.Figure or matplotlib.figure.Figure
        Interactive Plotly 3D scatter if Plotly is available, else matplotlib.
    """
    cloud = np.asarray(cloud, dtype=float)
    if cloud.ndim != 2 or cloud.shape[1] != 3:
        cloud = cloud.reshape(-1, 3) if cloud.size else np.zeros((0, 3))
    rt, mz, intensity = cloud[:, 0], cloud[:, 1], cloud[:, 2]
    i_max = float(intensity.max()) if len(intensity) else 1.0

    if title is None:
        title = f'PointIso point cloud ({len(cloud)} real peaks)'

    try:
        import plotly.graph_objects as go
        traces = []
        if sticks and len(cloud):
            # All sticks in one line trace: (mz, rt, 0) -> (mz, rt, I), split by NaN gaps.
            n = len(cloud)
            gap = np.full(n, np.nan)
            traces.append(go.Scatter3d(
                x=np.column_stack([mz, mz, gap]).ravel(),
                y=np.column_stack([rt, rt, gap]).ravel(),
                z=np.column_stack([np.zeros(n), intensity, gap]).ravel(),
                mode='lines',
                line=dict(width=3, color=np.repeat(intensity, 3),
                          colorscale='Viridis', cmin=0, cmax=i_max),
                hoverinfo='skip',
                showlegend=False,
            ))
        traces.append(go.Scatter3d(
            x=mz, y=rt, z=intensity,
            mode='markers',
            marker=dict(
                size=2 if sticks else 3,
                color=intensity,
                colorscale='Viridis',
                cmin=0, cmax=i_max,
                colorbar=dict(title='intensity'),
                opacity=0.85,
            ),
            hovertemplate='m/z: %{x:.4f}<br>RT: %{y:.3f} min<br>intensity: %{z:.0f}<extra></extra>',
            showlegend=False,
        ))
        fig = go.Figure(data=traces)
        fig.update_layout(
            title=title,
            scene=dict(
                xaxis_title='m/z',
                yaxis_title='Retention time (min)',
                zaxis=dict(title='intensity', rangemode='tozero'),
            ),
        )
        return fig
    except Exception:
        import logging
        logging.getLogger(__name__).warning(
            'Plotly unavailable for plot_lcms_point_cloud; falling back to matplotlib'
        )
        import matplotlib.pyplot as plt
        from mpl_toolkits.mplot3d import Axes3D  # noqa: F401  (registers 3d projection)
        fig = plt.figure(figsize=(8, 6))
        ax = fig.add_subplot(111, projection='3d')
        if len(cloud):
            if sticks:
                from mpl_toolkits.mplot3d.art3d import Line3DCollection
                segs = [[(x, y, 0.0), (x, y, z)] for x, y, z in zip(mz, rt, intensity)]
                ax.add_collection3d(Line3DCollection(
                    segs, colors=plt.get_cmap('viridis')(intensity / i_max), linewidths=1.5))
                ax.set_zlim(0, i_max * 1.05)
            sc = ax.scatter(mz, rt, intensity, c=intensity, cmap='viridis',
                            vmin=0, vmax=i_max, s=4 if sticks else 8)
            fig.colorbar(sc, ax=ax, shrink=0.6, label='intensity')
        ax.set_xlabel('m/z')
        ax.set_ylabel('Retention time (min)')
        ax.set_zlabel('intensity')
        ax.set_title(title)
        fig.tight_layout()
        return fig
```
